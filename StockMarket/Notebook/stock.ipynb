```python
dic={"apple":"AAPL","bitcoin":"BTC-USD","ethereum":"ETH-USD","google":"GOOGL","microsoft":"MSFT","amazon":"AMZN","meta":"META"}
```

```python
for key,value in dic.items():
    df=yf.download(value,period="max")
    df.columns=[col[0] for col in df.columns]
    df.reset_index(inplace=True)
    print(df.head())
    dic[key]=df
```

```python
import talib
for key,value in dic.items():
    value['SMA']=talib.SMA(value['Close'],timeperiod=14)
    value['EMA']=talib.EMA(value['Close'],timeperiod=14)
    value['upper_band'], value['middle_band'], value['lower_band'] = talib.BBANDS(value['Close'], timeperiod=14)
    value['slowk'], value['slowd'] = talib.STOCH(
        value['High'], 
        value['Low'], 
        value['Close'], 
        fastk_period=14, 
        slowk_period=3, 
        slowk_matype=0, 
        slowd_period=3, 
        slowd_matype=0
    )
    value['RSI']=talib.RSI(value['Close'],timeperiod=14)
    value['MACD'], value['MACD_signal'], value['MACD_hist'] = talib.MACD(value['Close'], fastperiod=12, slowperiod=26, signalperiod=9)
    value['ADX']=talib.ADX(value['High'], value['Low'], value['Close'], timeperiod=14)
    value['CCI']=talib.CCI(value['High'], value['Low'], value['Close'], timeperiod=14)
    value['Williams %R']=talib.WILLR(value['High'], value['Low'], value['Close'], timeperiod=14)
    value['ATR']=talib.ATR(value['High'], value['Low'], value['Close'], timeperiod=14)
    value['OBV']=talib.OBV(value['Close'], value['Volume'])
    value['MFI']=talib.MFI(value['High'], value['Low'], value['Close'], value['Volume'], timeperiod=14)
    value['SAR']=talib.SAR(value['High'], value['Low'], acceleration=0, maximum=0)
    value['TRIX']=talib.TRIX(value['Close'], timeperiod=30)
    value['ULTOSC']=talib.ULTOSC(value['High'], value['Low'], value['Close'], timeperiod1=7, timeperiod2=14, timeperiod3=28)
    value['PPO'] = talib.PPO(value['Close'], fastperiod=12, slowperiod=26, matype=0)
    value['KAMA']=talib.KAMA(value['Close'], timeperiod=30)
    # DPO is not available in talib, so we implement it manually
    dpo_period = 20
    dpo_shift = int(dpo_period / 2 + 1)
    dpo_ma = value['Close'].rolling(window=dpo_period, min_periods=0).mean()
    value['DPO'] = value['Close'] - dpo_ma.shift(dpo_shift)
    value['MOM']=talib.MOM(value['Close'], timeperiod=10)
    value['BOP']=talib.BOP(value['Open'], value['High'], value['Low'], value['Close'])
    value['HT_TRENDLINE']=talib.HT_TRENDLINE(value['Close'])
    value['HT_DCPERIOD']=talib.HT_DCPERIOD(value['Close'])
    value['HT_DCPHASE']=talib.HT_DCPHASE(value['Close'])
    value['HT_PHASOR_INPHASE'], value['HT_PHASOR_QUADRATURE'] = talib.HT_PHASOR(value['Close'])
    value['HT_SINE_SINE'], value['HT_SINE_LEADSINE'] = talib.HT_SINE(value['Close'])
    value['HT_TRENDMODE']=talib.HT_TRENDMODE(value['Close'])
    print(value.head())
```

```python
from xgboost import XGBRegressor, XGBClassifier
from sklearn.model_selection import RandomizedSearchCV, KFold
import os
from sklearn.preprocessing import StandardScaler
os.chdir('/workspaces/MLProject/StockMarket/Notebook')
model_path = os.path.join(os.getcwd(), 'all_models')
scaler_path = os.path.join(os.getcwd(), 'all_scalers')
l1=["Close_","Open_","High_","Low_","Volume_","trend"]
for key,value in dic.items():
    X=value.drop(columns=l1)
    y1=value['Close_']
    y2=value['Open_']
    y3=value['High_']
    y4=value['Low_']
    y5=value['Volume_']
    print(X.shape,y1.shape,y2.shape,y3.shape,y4.shape,y5.shape)
    model1=XGBRegressor()
    model2=XGBRegressor()
    model3=XGBRegressor()
    model4=XGBRegressor()
    model5=XGBRegressor()
    ssx=StandardScaler()
    X=ssx.fit_transform(X)
    ss1=StandardScaler()
    y1=ss1.fit_transform(y1.values.reshape(-1,1)).ravel()
    ss2=StandardScaler()
    y2=ss2.fit_transform(y2.values.reshape(-1,1)).ravel()
    ss3=StandardScaler()
    y3=ss3.fit_transform(y3.values.reshape(-1,1)).ravel()
    ss4=StandardScaler()
    y4=ss4.fit_transform(y4.values.reshape(-1,1)).ravel()
    ss5=StandardScaler()
    y5=ss5.fit_transform(y5.values.reshape(-1,1)).ravel()
    os.makedirs('scalers', exist_ok=True)
    os.chdir('scalers')
    import joblib
    joblib.dump(ssx, os.path.join(scaler_path, f'{key}_X_scaler.pkl'))
    joblib.dump(ss1, os.path.join(scaler_path, f'{key}_y1_scaler.pkl'))
    joblib.dump(ss2, os.path.join(scaler_path, f'{key}_y2_scaler.pkl'))
    joblib.dump(ss3, os.path.join(scaler_path, f'{key}_y3_scaler.pkl'))
    joblib.dump(ss4, os.path.join(scaler_path, f'{key}_y4_scaler.pkl'))
    joblib.dump(ss5, os.path.join(scaler_path, f'{key}_y5_scaler.pkl'))

    params={
        'n_estimators':[50,100,200],
        'learning_rate':[0.01,0.05,0.1,0.2],
        'max_depth':[3,5,7,9],
        'subsample':[0.6,0.8,1.0],
        'colsample_bytree':[0.6,0.8,1.0],
        'gamma':[0,0.1,0.2,0.3],
        'reg_alpha':[0,0.01,0.1,1],
        'reg_lambda':[1,1.5,2,3],
        'min_child_weight':[1,3,5,7]
    }
    kf=KFold(n_splits=5,shuffle=True,random_state=42)
    rs1=RandomizedSearchCV(model1,params,cv=kf,n_iter=10,scoring='neg_mean_squared_error',n_jobs=-1,verbose=1)
    rs2=RandomizedSearchCV(model2,params,cv=kf,n_iter=10,scoring='neg_mean_squared_error',n_jobs=-1,verbose=1)
    rs3=RandomizedSearchCV(model3,params,cv=kf,n_iter=10,scoring='neg_mean_squared_error',n_jobs=-1,verbose=1)
    rs4=RandomizedSearchCV(model4,params,cv=kf,n_iter=10,scoring='neg_mean_squared_error',n_jobs=-1,verbose=1)
    rs5=RandomizedSearchCV(model5,params,cv=kf,n_iter=10,scoring='neg_mean_squared_error',n_jobs=-1,verbose=1)
    rs1.fit(X,y1)
    rs2.fit(X,y2)
    rs3.fit(X,y3)
    rs4.fit(X,y4)
    rs5.fit(X,y5)
    print(f'Best parameters for Close_: {rs1.best_params_},accuracy:{-rs1.best_score_}')
    print(f'Best parameters for Open_: {rs2.best_params_}, accuracy:{-rs2.best_score_}')
    print(f'Best parameters for High_: {rs3.best_params_},accuracy:{-rs3.best_score_}')
    print(f'Best parameters for Low_: {rs4.best_params_},accuracy:{-rs4.best_score_}')
    print(f'Best parameters for Volume_: {rs5.best_params_},accuracy:{-rs5.best_score_}')
    model1=rs1.best_estimator_
    model2=rs2.best_estimator_
    model3=rs3.best_estimator_
    model4=rs4.best_estimator_
    model5=rs5.best_estimator_
    model1.fit(X,y1)
    model2.fit(X,y2)
    model3.fit(X,y3)
    model4.fit(X,y4)
    model5.fit(X,y5)
    print(f'Models trained for {key.capitalize()}')

    model1.save_model(os.path.join(model_path, f'{key}_close_model.json'))
    model2.save_model(os.path.join(model_path, f'{key}_open_model.json'))
    model3.save_model(os.path.join(model_path, f'{key}_high_model.json'))
    model4.save_model(os.path.join(model_path, f'{key}_low_model.json'))
    model5.save_model(os.path.join(model_path, f'{key}_volume_model.json'))
```
